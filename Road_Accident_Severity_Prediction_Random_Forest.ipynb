# ROAD ACCIDENT SEVERITY PREDICTION

## Using Random Forest Machine Learning Algorithm

### ABSTRACT

Road Accident Severity Prediction Using Random Forest is a machine learning project that predicts the severity level of a road accident from accident-related attributes such as weather condition, road condition, light condition, vehicle type, speed-related factors, and other relevant features.

The system uses preprocessing techniques to prepare categorical and numerical data and applies the Random Forest classification algorithm to learn patterns from historical accident records. The dataset is divided into training and testing sets, and the trained model is evaluated using accuracy, precision, recall, F1-score, and a confusion matrix.

The trained model can then predict the severity of a new accident as a suitable severity class. This project demonstrates how machine learning can support road-safety analysis and help identify accident severity patterns.

## PROGRAM

### 1. Import Required Libraries

````python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
````

### 2. Load Dataset

````python
df = pd.read_csv("road_accident.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
df.head()
````

### 3. Remove Rows with Missing Target Values

````python
df = df.dropna(subset=["Severity"])

print("Rows after removing missing target values:", df.shape[0])
````

### 4. Separate Input and Output

````python
X = df.drop("Severity", axis=1)
y = df["Severity"]

print("Input shape:", X.shape)
print("Target shape:", y.shape)
````

### 5. Identify Categorical and Numerical Columns

````python
categorical_cols = X.select_dtypes(include=["object"]).columns
numerical_cols = X.select_dtypes(exclude=["object"]).columns

print("Categorical columns:")
print(list(categorical_cols))

print("\nNumerical columns:")
print(list(numerical_cols))
````

### 6. Preprocessing

````python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        )
    ],
    remainder="passthrough"
)

print("Preprocessor created successfully!")
````

### 7. Random Forest Model

````python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

print("Random Forest model created!")
````

### 8. Create Pipeline

````python
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", model)
])

print("Pipeline created successfully!")
````

### 9. Train-Test Split

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
````

### 10. Train Model

````python
pipeline.fit(X_train, y_train)

print("Model Training Completed!")
````

### 11. Prediction

````python
y_pred = pipeline.predict(X_test)

print("Predictions generated successfully!")
print("First 10 predictions:", y_pred[:10])
````

### 12. Evaluation

````python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
````

### 13. Predict Severity for a New Accident

````python
new_accident = X_test.iloc[[0]]

prediction = pipeline.predict(new_accident)

print("New accident sample:")
display(new_accident)

print("\nPredicted Accident Severity:", prediction[0])
````
